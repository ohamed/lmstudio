```python
import pandas as pd
```

```python
# Load the CSV file
df = pd.read_csv('data/arxiv_csml_body_abstract.csv')
```

```python
# Sample 50 articles randomly (fixed seed for reproducibility)
sample_df = df.sample(n=50, random_state=42)
```

```python
# Save sample to new CSV with same headers, no index
sample_df.to_csv('data/arxiv_csml_body_abstract_S3.csv', index=False)

print("Sample of 50 articles saved successfully to 'arxiv_csml_body_abstract_sample.csv'")
```
