```python
import pandas as pd
import os
```

```python
# Load the CSV file
df = pd.read_csv('data/arxiv_csml_body_abstract.csv')
```

```python
# Sample 50 articles randomly (fixed seed for reproducibility)
sample_df = df.sample(n=50, random_state=42)
```

```python
file_path = 'data/arxiv_csml_body_abstract_S3.csv'
s3_found=False
if os.path.exists(file_path):
    s3_found=True
print(s3_found)
```

```python
if s3_found==False:
    # Save sample to new CSV with same headers, no index
    sample_df.to_csv('data/arxiv_csml_body_abstract_S3.csv', index=False)
    print("Sample of 50 articles saved successfully to 'arxiv_csml_body_abstract_sample.csv'")
```

### Create a sample of 100 for the ablation study:

```python
# load datasets
df_50 = pd.read_csv("data/arxiv_csml_body_abstract_S3.csv")
df_full = pd.read_csv("data/arxiv_csml_body_abstract.csv")
```

```python
file_path_100 = 'data/arxiv_csml_body_abstract_100.csv'
```

```python
# remove overlap using paper_id
df_new = df_full[~df_full["paper_id"].isin(df_50["paper_id"])]
```

```python
# sample 50 new articles
df_new_50 = df_new.sample(n=50, random_state=42)
```

```python
# combine to get 100 articles
df_100 = pd.concat([df_50, df_new_50], ignore_index=True)
```

```python
# Save the 100 sample to file system
df_100.to_csv(file_path_100, index=False)
```
