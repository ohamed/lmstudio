```python
import pandas as pd
import numpy as np
```

```python
#metrics = ["BLEU", "ROUGE-1-F1", "ROUGE-L-F1", "BERTScore-F1"]

## Double-check the names in target DF.
metrics = ["BLEU", "ROUGE-1", "ROUGE-L", "BERTScore"]
```

```python
def reportMetrics(df):
    for metric in metrics:
        mean = df[metric].mean()
        sd = df[metric].std()
        
        print(f"{metric}: {mean:.2f} ± {sd:.2f}")
```
